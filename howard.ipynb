## Import modules

````python
# Import
import os
from howard.objects.variants import Variants

````

## Create and load input VCF

````python
# Input VCF
input_vcf = os.path.join("tests", "data", "example.vcf.gz")

# Create variants object with input VCF and load data
variants = Variants(input=input_vcf, load=True)
````

## Stats on variants

````python
# Get stats
stats = variants.get_stats()

# Print stats such as number of variants, samples, INFO fields, FORMAT fields
print(stats.get("Infos", {}))
````

## Query variants

````python
# Query
query = """
SELECT *
FROM variants
"""

# Execute query and retrieve dataframe
df = variants.get_query_to_df(query=query)

# Print dataframe
print(df)
````

## Query variants INFO field

````python
# Query
query = """
SELECT INFO
FROM variants
"""

# Execute query and retrieve dataframe
df = variants.get_query_to_df(query=query)

# Print dataframe
print(df)
````

## Annotate variants with a parquet database (dbNSFP)

````python
# Database
database = os.path.join("tests", "databases", "annotations", "current", "hg19", "dbnsfp42a.parquet")

# Parameters
param = {
    "annotation": {
        "parquet": {
            "annotations": {
                database: {}
            }
        }
    }
}

# Load parameters
variants.set_param(param=param)

# Annotate
variants.annotation_parquet()

````

## Use Variants View

````python
# Create view
variants.create_annotations_view(table="variants", view="variants_view", view_type="view", view_mode="explore", info_prefix_column="", info_struct_column="INFOS", sample_struct_column="SAMPLES")

# Query
query = """
SELECT "#CHROM", POS, REF, ALT, INFOS.Polyphen2_HVAR_pred AS PolyPhen_pred, INFOS.Polyphen2_HVAR_score AS PolyPhen_score, SAMPLES.sample1.GT, SAMPLES.sample1.DP
FROM variants_view
"""

# Execute query and retrieve dataframe
df = variants.get_query_to_df(query=query)

# Print dataframe
print(df)
````
